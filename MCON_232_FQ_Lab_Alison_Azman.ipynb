Linear Search- iterative versio

```python
#to searcg the entire space for one element's value

def linear_search_itr(nlist:list, target:int):
  for  index, value in enumerate(nlist):
    if value==target:
      print(f"Item found at index {index} value {value}")
      return True

  return False

print(linear_search_itr([20,30,40,50], 50))
print(linear_search_itr([20,30,40,50], 35))
```

Recursive Linear search
- base case
- recursive case
  - which subdivides the problem space

```python
def linear_search_rec(nlist:list, target:int,current_index=0):

  #Base Case
  if current_index==len(nlist):
    return False

  #second base case
  if nlist[current_index]==target:
    print(f"Item found at index {current_index} value {nlist[current_index]}")
    return True

  else:
    return linear_search_rec(nlist, target, current_index+1)



print(linear_search_rec([20,30,40,50], 50,0))
print(linear_search_rec([20,30,40,50], 35,0))
```

Binary Search Example:
This works on sorted data only

This is the logn split

take half of the data and keep taking half of the data

Exclude the midpoint as well

```python
def binary_search(nums:list, target:int, low:int, high:int):

  #base case 1
  if low>high:
    return -1

  mid= (low+high)//2
  #base case 2
  if nums[mid]==target:
    return mid

  #recursive cases
  #If the target is less than the midpoint
  elif target<nums[mid]:
    return binary_search(nums, target, low, mid-1)

  #If the target is greater than the midpoint
  else:
    return binary_search(nums, target, mid+1, high)

nlist=[10,20,30,40,50]
print(binary_search(nlist, 100,0, len(nlist)-1))
```
