```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
  total_vehicles = 0

  def __init__(self, color:str, speed:int):
    self.color=color
    self.speed=speed
    Vehicle.total_vehicles+=1


  @abstractmethod
  def honk(self):
    pass

  @classmethod
  def get_number_of_vehicles(cls):
    return cls.total_vehicles

  @staticmethod
  def is_safe_speed(speed:int):
    if speed>25 and speed<50:
      return True

    else:
      return False



class Car(Vehicle):

  def __init__(self, color: str, speed: int, seat_amt: int):
        super().__init__(color, speed)
        self.seat_amt = seat_amt

  def honk(self):
     return f"The {self.color} car is beeping."


class Truck(Vehicle):

  def __init__(self, color: str, speed: int, wheel_amt: int):
        super().__init__(color, speed)
        self.wheel_amt = wheel_amt

  def honk(self):
    return f"The {self.color} truck is HOOOOONNNKKING."


class Motorcycle(Vehicle):

  def __init__(self, color: str, speed: int):
        super().__init__(color, speed)

  def honk(self):
    return f"The {self.color} motorcycle is meeping."




car = Car("black", 34, 4)
truck= Truck("white", 76, 18)
motorcycle=Motorcycle("pink", 40)

vehicles=[car, truck, motorcycle]

for vehicle in vehicles:
  print(vehicle.honk())


print(Vehicle.get_number_of_vehicles())

print(Vehicle.is_safe_speed(30))
```
